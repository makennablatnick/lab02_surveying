##### Grid Setup
	- First, we put down a stake at our first corner
	- From there, we used the compass app on a smartphone to orient the first edge pointed North
	- The grid was set up using a triangle method using Pythagreon's Theorum to ensure that it was perfectly square
	- After placing the first stake, we walked the tape measure out 20 meters and then walked another tape measure 28.82 at a 45 degree from the first edge (calculated using Pythagorean’s theorem) meters out to make the square. 
	- We placed a flag every 5 meters along the 20 meter mark and moved the tape measure in another 5 meters to get the rows and collumns lined up
	- It is worth noting that we placed a stake at each corner of the square 
	- The flags were placed in a color order starting at the starting stake and moving to the West
		- Blue (A), Yellow (B), Pink (C), Red (D), Green (E)
	- The grid was not necessarily centered at a specific location. It is simply in a spot where the entire square could fit onto Kafadar and where every point remained in sight of the base station.
	- The grid sits in between the known control points, and we believe that it ended up being a good choice
	- Since the grid was oriented North, it was easy to sanity check our edges by lining up the compass app to see if each edge pointed in a cardinal direction.
	- Things mostly went to plan, but we did end up with a corner that ended up being rougly 5 cm off of where it theoretically should have been due to human error.
