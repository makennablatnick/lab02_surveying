##### Grid Setup

	- The grid was set up using a triangle method to ensure that it was perfectly square
	- We placed a stake at the Northernmost point, walked the tape measure out 20 meters and then walked another tape measure 28.82 (calculated using Pythagorean’s theorem) meters out to make the square. 
	- We placed a mark every 5 feet along the 20 meter mark and moved the tape measure in another 5 meters to get the rows and collumns lined up 
	- The flags were placed in a color order starting at the starting stake and moving to the left
		- Blue, Light Green, Pink, Orange, Dark Green
